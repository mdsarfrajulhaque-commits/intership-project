```python


from PIL import Image, ImageDraw, ImageFont
from IPython.display import display

print("=== Welcome to Supermarket Billing ===")

total_bill = 0
item_list = []


while True:
    item_name = input("Enter item name (type 'done' to finish): ")

    if item_name.lower() == 'done':
        break

    try:
        item_price = float(input(f"Enter price for {item_name}: ₹"))

        if item_price < 0:
            print("Price cannot be negative!\n")
            continue

        total_bill += item_price
        item_list.append((item_name, item_price))

        print(f"Added {item_name} to cart.\n")

    except ValueError:
        print("Invalid price! Enter a valid number.\n")



width = 600
height = 350 + len(item_list) * 45

img = Image.new("RGB", (width, height), "#f5f5f5")
draw = ImageDraw.Draw(img)


try:
    title_font = ImageFont.truetype("DejaVuSans-Bold.ttf", 28)
    font = ImageFont.truetype("DejaVuSans.ttf", 20)
    bold_font = ImageFont.truetype("DejaVuSans-Bold.ttf", 22)
except:
    title_font = font = bold_font = ImageFont.load_default()


draw.rectangle((20, 20, 580, height - 20),
               fill="white", outline="black", width=3)

draw.text((75, 40), "SUPER MARKET BILL",
          fill="black", font=title_font)

draw.text((210, 90), "FINAL RECEIPT",
          fill="green", font=bold_font)

draw.line((45, 135, 555, 135), fill="black", width=2)

y = 155

for name, price in item_list:
    draw.text((55, y), name, fill="black", font=font)
    draw.text((420, y), f"Rs. {price:.2f}",
              fill="black", font=font)
    y += 45

draw.line((45, y, 555, y), fill="black", width=2)
y += 20


draw.text((55, y), "TOTAL AMOUNT:",
          fill="black", font=bold_font)

draw.text((350, y), f"Rs. {total_bill:.2f}",
          fill="green", font=bold_font)

y += 55

draw.text((120, y), "Thank You For Shopping!",
          fill="blue", font=bold_font)


display(img)


img.save("Supermarket_Receipt.png")

print("\nReceipt image saved successfully!")
```
